```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# -------------------------------
# STYLE (APPLY FIRST)
# -------------------------------
plt.rcParams.update({
    "figure.dpi": 800,
    "savefig.dpi": 800,
    "savefig.bbox": "tight",
    "font.family": "serif",
    "font.serif": "DejaVu Serif",
    "font.size": 20,
    "axes.titlesize": 20,
    "axes.labelsize": 20,
    "xtick.labelsize": 20,
    "ytick.labelsize": 20,
    "legend.fontsize": 18,
    "axes.grid": True,
    "grid.alpha": 0.25,
    "grid.linestyle": "--",
    "axes.labelweight": "bold",
    "font.weight": "bold",
})

# -------------------------------
# LOAD ALL MODALITIES
# -------------------------------
df_data = pd.read_csv("family_drift_results_data.csv")
df_json = pd.read_csv("family_drift_results_json.csv")
df_gml  = pd.read_csv("family_drift_results_gml.csv")

df_data["modality"] = "data"
df_json["modality"] = "json"
df_gml["modality"]  = "gml"

# Combine
df = pd.concat([df_data, df_json, df_gml], ignore_index=True)

# -------------------------------
# PREPARE DATA
# -------------------------------
max_drift = df["drift"].max()
df["drift_norm"] = df["drift"] / max_drift if max_drift != 0 else 0
df["stability"] = 1 - df["drift_norm"]

# -------------------------------
# SELECT FAMILIES
# -------------------------------
top_families = [
    "dowgin", "airpush", "kuguo", "smsreg",
    "dnotua", "gappusin", "revmob",
    "leadbolt", "adwo", "youmi"
]

df_plot = df[df["family"].isin(top_families)]

# -------------------------------
# BUILD GROUPED DATA
# -------------------------------
modalities = ["data", "json", "gml"]

data = []
positions = []
labels = []

pos = 1
width = 0.25

for fam in top_families:
    for i, mod in enumerate(modalities):
        vals = df_plot[
            (df_plot["family"] == fam) &
            (df_plot["modality"] == mod)
        ]["stability"].dropna().values

        data.append(vals)
        positions.append(pos + i * width)

    labels.append(fam)
    pos += 1 + width * 3

# -------------------------------
# PLOT
# -------------------------------
plt.figure(figsize=(16, 6))

box = plt.boxplot(
    data,
    positions=positions,
    widths=0.2,
    patch_artist=True,
    showfliers=True
)

# Colors for modalities
colors = ["#1f77b4", "#ff7f0e", "#2ca02c"]  # data, json, gml

for i, patch in enumerate(box["boxes"]):
    patch.set_facecolor(colors[i % 3])
    patch.set_alpha(0.7)

# Center x-ticks per family
group_centers = [
    np.mean(positions[i*3:(i+1)*3])
    for i in range(len(top_families))
]

plt.xticks(group_centers, labels, rotation=25, ha="right")

plt.ylabel("Stability Score")
plt.xlabel("Malware Family")

# Legend
from matplotlib.patches import Patch
legend_elements = [
    Patch(facecolor=colors[0], label="Static"),
    Patch(facecolor=colors[1], label="Dynamic"),
    Patch(facecolor=colors[2], label="Graph-based"),
]
plt.legend(handles=legend_elements)

plt.grid(axis="y", linestyle="--", alpha=0.4)

plt.tight_layout()

# -------------------------------
# SAVE
# -------------------------------

import matplot2tikz
matplot2tikz.save("test.tex")
plt.savefig("family_stability_grouped.pdf", format="pdf", bbox_inches="tight")
plt.show()
```
